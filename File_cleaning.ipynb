```python
#Import the libaries 
import pandas as pd
import numpy as np
```

```python
#Read the file 
online_retail_raw=pd.read_csv("/Users/jahaankachwala/Downloads/Projects /Online Retail/Online_Retail.csv")
online_shoppers_raw=pd.read_csv("/Users/jahaankachwala/Downloads/Projects /Online Retail/online_shoppers_intention.csv")
```

```python
#Cleaning online_retail raw

#Check the data
print(online_retail_raw.head())
print(online_retail_raw.info())

#Remove duplicate rows
df = online_retail_raw.drop_duplicates()

#Remove rows that are completely empty
df = online_retail_raw.dropna(how="all")

# Fill missing numerical values with the median
for col in online_retail_raw.select_dtypes(include="number").columns:
    df[col] = df[col].fillna(df[col].median())

#Fill missing text values with "Unknown"
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].fillna("Unknown")

#Remove extra spaces from text
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()

#Save the cleaned file
df.to_csv("online_retail_cleaned.csv", index=False)

#Check the result
print("Cleaning complete!")
print("Rows:", len(df))
print("Missing values:", online_retail_raw.isnull().sum().sum())

```
